```python
from datasets import load_dataset
from itertools import islice
```

# dataset loading

```python
dataset_multi = load_dataset(
       "joelito/Multi_Legal_Pile",
       name="en_all",              # language config: en, de, fr, etc.
       split="train",
       streaming=True,
       trust_remote_code=True
   )
```

```python
dataset_pile = load_dataset(
       "pile-of-law/pile-of-law",
       name="r_legaladvice",   # required config name
       split="train",
       streaming=True
   )
```

```python
dataset_eur = load_dataset("lex_glue", "eurlex",split="train")
```

```python
dataset_qa = load_dataset("alzoubi36/policy_qa", split="train",streaming=True)
```

```python
sample_docs_multi = list(islice(dataset_multi, 1000))
print("multi done")
sample_docs_pile = list(islice(dataset_pile, 1000))
print("pile done")
sample_docs_eur = list(islice(dataset_eur, 1000))
print("eur done")
```

```python
sample_docs_qa = list(islice(dataset_qa, 1000))
print("qa done")
```

# data converting to txt files

```python
import os
from tqdm import tqdm
```

```python
output_dir = "../data/raw/Multi_Legal_Pile"
os.makedirs(output_dir, exist_ok=True)

for i, row in tqdm(enumerate(sample_docs_multi), total=len(sample_docs_multi)):
    text = row.get("text")

    if text and text.strip():
        with open(f"{output_dir}/doc_{i}.txt", "w", encoding="utf-8") as f:
            f.write(text)
```

# metdata extraction

```python
def extract_metadata(samples, source_name, language="en", doc_type="legal"):
    rows = []
    
    for idx, sample in enumerate(samples):
        text = sample.get("text", "")
        
        row = {
            "doc_id": f"{source_name}_{idx}",
            "source": source_name,
            "language": language,
            "document_type": doc_type,
            "text_length": len(text),
        }
        
        rows.append(row)
    
    return rows
```

```python
rows_pile = extract_metadata(sample_docs_pile, "pile_of_law")
rows_eur = extract_metadata(sample_docs_eur, "eur")
rows_qa = extract_metadata(sample_docs_qa, "policy_qa")
rows_multi = extract_metadata(sample_docs_multi, "multi_legal_pile")
```

```python
total_rows= rows_eur +rows_qa +rows_multi +rows_pile
```

```python
import pandas as pd
df = pd.DataFrame(total_rows)
df.to_csv("../data/metadata.csv", index=False)
```
